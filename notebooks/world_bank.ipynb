```python
from pathlib import Path
import sys
import pandas as pd


# Project root
PROJECT_ROOT = Path.cwd().parent


# Allow imports from project root
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))


# Import file paths
from python.utils.file_paths import (
    ENERGY_IMPORTS,
    ENERGY_USE,
    GDP_CURRENT_USD,
    GDP_GROWTH,
    GDP_PER_CAPITA,
    IMPORTS,
    EXPORTS,
    INFLATION,
    WORLD_BANK_PROCESSED_FOLDER
)
```

```python
WORLD_BANK_FILES = {
    "energy_imports": ENERGY_IMPORTS,
    "energy_use": ENERGY_USE,
    "gdp_current_usd": GDP_CURRENT_USD,
    "gdp_growth": GDP_GROWTH,
    "gdp_per_capita": GDP_PER_CAPITA,
    "imports": IMPORTS,
    "exports": EXPORTS,
    "inflation": INFLATION,
}

WORLD_BANK_FILES
```

```python
for name, path in WORLD_BANK_FILES.items():

    print("=" * 80)
    print(name.upper())
    print("=" * 80)
    print("Path:", path)
    print("Exists:", path.exists())

    if path.exists():
        print("Size:", path.stat().st_size, "bytes")

        with open(path, "r", encoding="utf-8") as f:
            for _ in range(3):
                print(f.readline().strip())

    print()
```

```python
# ============================================================
# LOAD RAW WORLD BANK DATA
# ============================================================

raw_data = {}

for name, path in WORLD_BANK_FILES.items():

    df = pd.read_csv(path)

    raw_data[name] = df

    print(f"{name:20} -> {df.shape}")
```

```python
# ============================================================
# CHECK COLUMNS
# ============================================================

for name, df in raw_data.items():

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print(df.columns.tolist())
    print()
```

```python
# ============================================================
# DATA TYPES
# ============================================================

for name, df in raw_data.items():

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print(df.dtypes)
    print()
```

```python
# ============================================================
# MISSING VALUES
# ============================================================

for name, df in raw_data.items():

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    missing = df.isna().sum()

    missing_pct = (
        df.isna().mean() * 100
    ).round(2)

    profile = pd.DataFrame({
        "missing_count": missing,
        "missing_pct": missing_pct
    })

    print(profile)
    print()
```

```python
# ============================================================
# COUNTRY COVERAGE
# ============================================================

EXPECTED_COUNTRIES = {
    "IND",
    "CHN",
    "JPN",
    "KOR",
    "SGP",
    "USA",
    "DEU",
    "GBR",
    "FRA",
    "CAN",
    "AUS",
    "BRA",
    "SAU",
    "ARE",
}

for name, df in raw_data.items():

    countries = set(
        df["countryiso3code"]
        .dropna()
        .unique()
    )

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print("Countries found:", len(countries))
    print("Missing expected:", EXPECTED_COUNTRIES - countries)
    print("Unexpected:", countries - EXPECTED_COUNTRIES)
    print()
```

```python
# ============================================================
# YEAR COVERAGE
# ============================================================

for name, df in raw_data.items():

    years = pd.to_numeric(
        df["year"],
        errors="coerce"
    )

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print("Minimum year:", years.min())
    print("Maximum year:", years.max())
    print("Unique years:", years.nunique())
    print()
```

```python
# ============================================================
# COUNTRY-YEAR COVERAGE
# ============================================================

for name, df in raw_data.items():

    temp = df.copy()

    temp["year"] = pd.to_numeric(
        temp["year"],
        errors="coerce"
    )

    coverage = (
        temp.groupby("countryiso3code")["year"]
        .agg(
            min_year="min",
            max_year="max",
            years_available="nunique"
        )
        .sort_index()
    )

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print(coverage)
    print()
```

```python
# ============================================================
# DUPLICATE COUNTRY-YEAR CHECK
# ============================================================

for name, df in raw_data.items():

    duplicates = (
        df.groupby(
            ["countryiso3code", "year"]
        )
        .size()
        .reset_index(name="count")
    )

    duplicates = duplicates[
        duplicates["count"] > 1
    ]

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print("Duplicate country-year combinations:", len(duplicates))

    if len(duplicates) > 0:
        print(duplicates)

    print()
```

```python
# ============================================================
# MISSING VALUE LOCATIONS
# ============================================================

for name, df in raw_data.items():

    missing = df[df["value"].isna()][
        [
            "countryiso3code",
            "country",
            "year",
            "value"
        ]
    ]

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    if missing.empty:
        print("No missing values.")
    else:
        print(missing.to_string(index=False))

    print()
```

```python
# ============================================================
# MISSING VALUES BY COUNTRY
# ============================================================

for name, df in raw_data.items():

    missing_by_country = (
        df["value"].isna()
        .groupby(df["countryiso3code"])
        .sum()
        .sort_values(ascending=False)
    )

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print(missing_by_country[missing_by_country > 0])
    print()
```

```python
# ============================================================
# MISSING VALUES BY YEAR
# ============================================================

for name, df in raw_data.items():

    missing_by_year = (
        df["value"].isna()
        .groupby(df["year"])
        .sum()
    )

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print(missing_by_year[missing_by_year > 0])
    print()
```

```python
# ============================================================
# CLEAN WORLD BANK DATA
# ============================================================

processed_data = {}

for name, df in raw_data.items():

    clean_df = df.copy()

    # Keep only analytical columns
    clean_df = clean_df[
        [
            "countryiso3code",
            "country",
            "year",
            "value"
        ]
    ]

    # Rename columns
    clean_df = clean_df.rename(
        columns={
            "countryiso3code": "country_code"
        }
    )

    # Data types
    clean_df["year"] = clean_df["year"].astype("int64")
    clean_df["value"] = pd.to_numeric(
        clean_df["value"],
        errors="coerce"
    )

    # Sort
    clean_df = clean_df.sort_values(
        ["country_code", "year"]
    ).reset_index(drop=True)

    processed_data[name] = clean_df

    print("=" * 70)
    print(name.upper())
    print("=" * 70)
    print(clean_df.dtypes)
    print(clean_df.shape)
    print()
```

```python
# ============================================================
# VALIDATE CLEANED WORLD BANK DATA
# ============================================================

EXPECTED_COUNTRIES = {
    "ARE", "AUS", "BRA", "CAN", "CHN", "DEU", "FRA",
    "GBR", "IND", "JPN", "KOR", "SAU", "SGP", "USA"
}

for name, df in processed_data.items():

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    # --------------------------------------------------------
    # 1. Required-column nulls
    # --------------------------------------------------------
    required_nulls = df[
        ["country_code", "country", "year"]
    ].isna().sum()

    print("\nRequired column nulls:")
    print(required_nulls)

    # --------------------------------------------------------
    # 2. Duplicate country-year combinations
    # --------------------------------------------------------
    duplicates = df.duplicated(
        subset=["country_code", "year"]
    ).sum()

    print(f"\nDuplicate country-year combinations: {duplicates}")

    # --------------------------------------------------------
    # 3. Year range
    # --------------------------------------------------------
    print(f"\nMinimum year: {df['year'].min()}")
    print(f"Maximum year: {df['year'].max()}")
    print(f"Unique years: {df['year'].nunique()}")

    # --------------------------------------------------------
    # 4. Country coverage
    # --------------------------------------------------------
    countries_found = set(df["country_code"].dropna().unique())

    missing_countries = EXPECTED_COUNTRIES - countries_found
    unexpected_countries = countries_found - EXPECTED_COUNTRIES

    print(f"\nCountries found: {len(countries_found)}")
    print(f"Missing expected countries: {missing_countries}")
    print(f"Unexpected countries: {unexpected_countries}")

    # --------------------------------------------------------
    # 5. Numeric value check
    # --------------------------------------------------------
    print(
        f"\nMissing values in value column: "
        f"{df['value'].isna().sum()}"
    )

    # --------------------------------------------------------
    # 6. Negative values
    # --------------------------------------------------------
    negative_count = (df["value"] < 0).sum()

    print(f"Negative values: {negative_count}")

    if negative_count > 0:
        print("\nNegative-value examples:")
        print(
            df[df["value"] < 0]
            [["country_code", "year", "value"]]
            .head(10)
        )

    print()
```

```python
# ============================================================
# SAVE CLEANED WORLD BANK DATA
# ============================================================

WORLD_BANK_PROCESSED_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)

for name, df in processed_data.items():

    output_path = WORLD_BANK_PROCESSED_FOLDER / f"{name}.csv"

    df.to_csv(
        output_path,
        index=False
    )

    print(f"{name:20} -> {output_path}")
```
