```python
from pathlib import Path
import sys
import pandas as pd

# Project Root
PROJECT_ROOT = Path.cwd().parent

# Allow imports from project root
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

from python.utils.file_paths import (
    BRENT_CRUDE_PRICE,
    WTI_CRUDE_PRICE,
    FRED_PROCESSED_FOLDER
)
```

```python
# ============================================================
# LOAD RAW FRED DATA
# ============================================================

raw_data = {
    "brent_crude_price": pd.read_csv(BRENT_CRUDE_PRICE),
    "wti_crude_price": pd.read_csv(WTI_CRUDE_PRICE)
}

for name, df in raw_data.items():
    print(f"{name:20} -> {df.shape}")
```

```python
# ============================================================
# STRUCTURE INSPECTION
# ============================================================

for name, df in raw_data.items():

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData types:")
    print(df.dtypes)

    print("\nFirst 5 rows:")
    print(df.head())

    print("\nLast 5 rows:")
    print(df.tail())

    print()
```

```python
# ============================================================
# FRED DATA QUALITY INSPECTION
# ============================================================

for namme, df in raw_data.items():

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData types:")
    print(df.dtypes)

    print("\nFirst 5 rows:")
    print(df.head())

    print("\nLast 5 rows:")
    print(df.tail())

    print()
```

```python
print(raw_data.keys())

for name in raw_data:
    print(name)
```

```python
for name, df in raw_data.items():

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print("\nMissing marker ('.'):")
    print((df["value"] == ".").sum())

    print("\nActual nulls:")
    print(df.isna().sum())

    print("\nDate range:")
    print("Minimum:", df["date"].min())
    print("Maximum:", df["date"].max())

    print("\nDuplicate dates:")
    print(df["date"].duplicated().sum())

    print("\nUnique dates:")
    print(df["date"].nunique())
```

```python
# ============================================================
# INSPECT FRED MISSING OBSERVATION DATES
# ============================================================

for name, df in raw_data.items():

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    missing_dates = df.loc[df["value"] == ".", "date"]

    print(f"\nMissing observations: {len(missing_dates)}")

    print("\nFirst 30 missing dates:")
    print(missing_dates.head(30).to_list())

    print("\nLast 30 missing dates:")
    print(missing_dates.tail(30).to_list())
```

```python
# ============================================================
# COMPARE MISSING DATES
# ============================================================

brent_missing = set(
    raw_data["brent_crude_price"]
    .loc[raw_data["brent_crude_price"]["value"] == ".", "date"]
)

wti_missing = set(
    raw_data["wti_crude_price"]
    .loc[raw_data["wti_crude_price"]["value"] == ".", "date"]
)

print("Brent missing:", len(brent_missing))
print("WTI missing:", len(wti_missing))

print("\nMissing in both:", len(brent_missing & wti_missing))
print("Missing only in Brent:", len(brent_missing - wti_missing))
print("Missing only in WTI:", len(wti_missing - brent_missing))
```

```python
# ============================================================
# CLEAN FRED DATA
# ============================================================

processed_data = {}

for name, df in raw_data.items():

    clean_df = df.copy()

    # Convert date to datetime
    clean_df["date"] = pd.to_datetime(clean_df["date"])

    # Convert FRED missing marker "." to NaN
    clean_df["value"] = clean_df["value"].replace(".", pd.NA)

    # Convert price to numeric
    clean_df["value"] = pd.to_numeric(
        clean_df["value"],
        errors="coerce"
    )

    # Remove FRED vintage metadata
    clean_df = clean_df.drop(
        columns=["realtime_start", "realtime_end"]
    )

    processed_data[name] = clean_df

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print("\nColumns:")
    print(clean_df.columns.tolist())

    print("\nData types:")
    print(clean_df.dtypes)

    print("\nMissing values:")
    print(clean_df.isna().sum())

    print("\nShape:")
    print(clean_df.shape)
```

```python
# ============================================================
# FINAL VALIDATION — FRED PROCESSED DATA
# ============================================================

for name, df in processed_data.items():

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print("\nDuplicate dates:")
    print(df["date"].duplicated().sum())

    print("\nDate range:")
    print(f"{df['date'].min().date()} -> {df['date'].max().date()}")

    print("\nNon-numeric values:")
    print(df["value"].apply(lambda x: not pd.isna(x) and not isinstance(x, (int, float))).sum())

    print("\nNegative prices:")
    print((df["value"] < 0).sum())

    print("\nMissing values:")
    print(df.isna().sum())

    print()
```

```python
# ============================================================
# SAVE PROCESSED FRED DATA
# ============================================================

FRED_PROCESSED_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)

for name, df in processed_data.items():

    output_path = FRED_PROCESSED_FOLDER / f"{name}.csv"

    df.to_csv(
        output_path,
        index=False
    )

    print(f"{name:20} -> {output_path}")
```
